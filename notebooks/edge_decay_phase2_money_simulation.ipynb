## 採否と解釈
代表1.5%ではALL FinalCapital改善と継続OOS純利益改善を両方必須、他の3 Riskでも両差分>=0を必須と事前固定。すべて減少し不通過。DDやOOS Rの改善で救済しない。

## 取引明細の保存先変更（ユーザー指定）
計算完了後、公開GitHubへの詳細取引ログ送信が自動承認レビューで拒否された。ユーザーは「取引明細はローカル保存のみ」と指定したため、事前計画の計算仕様・結果を変更せず、明細CSVの公開保存だけを取り止めた。GitHubには7本の集計・検証・記録CSVを保存し、約56MBの取引明細CSVはユーザー向けローカルoutputsへ保存する。run recordのCSVHashesにはローカル明細のhashも含む。notebook実行時は/contentに明細を生成し、Drive保存は任意。


```python
from pathlib import Path
import hashlib, importlib.util
from urllib.request import urlopen
IMPLEMENTATION_SHA = 'f0e0641f10f27eb6b998ca24319de66d7470bee9'
MANIFEST = {'src/research/edge_decay_phase2_money_simulation.py': 'a74c8d378b1ea08150ebb70fe20fabcdc94acf8d1ad7e03c758f016b10cbd858', 'tests/verify_edge_decay_phase2_money_simulation.py': '5a3aa53443d9736c62eccf5838ec8613f8a446b39842862abe217b45aea92b17', 'src/portfolio_money_management_sim_v1_1.py': '62e878ff49e5692613e3f5e837faf8cf96fb7ca9851694fbf4f85f72f2a36848'}
code_dir = Path('/content/edge_decay_phase2_money_code')
code_dir.mkdir(parents=True, exist_ok=True)
for relative, expected in MANIFEST.items():
    raw = urlopen(f'https://raw.githubusercontent.com/TR-KJ/time-entry-portfolio-lab/{IMPLEMENTATION_SHA}/{relative}').read()
    assert hashlib.sha256(raw).hexdigest() == expected
    (code_dir / Path(relative).name).write_bytes(raw)
def load_module(name, filename):
    spec = importlib.util.spec_from_file_location(name, code_dir / filename)
    module = importlib.util.module_from_spec(spec)
    spec.loader.exec_module(module)
    return module
money = load_module('money', 'edge_decay_phase2_money_simulation.py')
audit = load_module('money_audit', 'verify_edge_decay_phase2_money_simulation.py')
```

```python
from google.colab import files
uploaded = files.upload()
matching = [name for name, raw in uploaded.items() if hashlib.sha256(raw).hexdigest() == money.BASELINE_HASH]
assert len(matching) == 1, '固定Baseline hashと一致するCSVを1本選択してください'
baseline_path = matching[0]
```

```python
output_dir = Path('/content')
money.run(baseline_path, output_dir, IMPLEMENTATION_SHA, execution='COLAB')
audit.verify(baseline_path, output_dir, code_dir / 'portfolio_money_management_sim_v1_1.py')
import pandas as pd
from IPython.display import display
for name in ('period_results', 'yearly_results', 'decision', 'constraint_status', 'verification', 'run_record'):
    print(name)
    display(pd.read_csv(output_dir / (money.PREFIX + name + '.csv')))
```
