```python
if PREPARE_ENVIRONMENT:
    import importlib.util, subprocess, sys
    from pathlib import Path
    try:
        in_colab = importlib.util.find_spec("google.colab") is not None
    except ModuleNotFoundError:
        in_colab = False
    if not in_colab:
        raise PermissionError("Prepare in Google Colab only")
    if len(STAGE6_FREEZE_SHA) != 40 or any(c not in "0123456789abcdef" for c in STAGE6_FREEZE_SHA):
        raise ValueError("Chat確認済みStage6 Freeze SHAを設定してください")
    subprocess.run([sys.executable,"-m","pip","install","numpy==2.3.5","pandas==2.2.3"],check=True)
    repo = Path(REPO_ROOT)
    if not repo.exists():
        subprocess.run(["git","clone","https://github.com/TR-KJ/time-entry-portfolio-lab.git",str(repo)],check=True)
    subprocess.run(["git","diff","--quiet","HEAD","--"],cwd=repo,check=True)
    subprocess.run(["git","fetch","origin",STAGE6_FREEZE_SHA],cwd=repo,check=True)
    subprocess.run(["git","checkout","--detach",STAGE6_FREEZE_SHA],cwd=repo,check=True)
    sys.path.insert(0,str(repo/"src/research"))
    import numpy as np, pandas as pd
    if (np.__version__,pd.__version__) != ("2.3.5","2.2.3"):
        raise RuntimeError("読込済みライブラリが旧版の場合はランタイムを再起動し、準備セルを再実行してください")
    from b6.stage6_search import verify_release
    verify_release(STAGE6_FREEZE_SHA)
    from b6.stage6_config import load_config
    if load_config() != FROZEN_CONFIG:
        raise ValueError("Notebook/config mismatch")
    print("Stage6 preparation PASS")
```

```python
if RUN_STAGE6_FULL:
    if not PREPARE_ENVIRONMENT or not CHAT_CONFIRMED_STAGE6_FREEZE:
        raise PermissionError("固定コード準備とChat確認が必要です")
    from b6.stage6_search import full_validation
    summary = full_validation(DATA_ROOT, OUTPUT_ROOT, STAGE6_FREEZE_SHA,
                              run_full=RUN_STAGE6_FULL, confirmed=CHAT_CONFIRMED_STAGE6_FREEZE)
else:
    print("Stage6 full Validation OFF。Chat確認後にColabで実行。")
```

```python
if RUN_STAGE6_FULL:
    from pathlib import Path
    out = Path(OUTPUT_ROOT)
    progress = json.loads((out/"progress.json").read_text())
    if progress["State"] != "COMPLETE_STAGE6_VALIDATION_ONLY" or progress["CompletedJobs"] != FROZEN_CONFIG["candidate_count"]:
        raise RuntimeError("50候補完走前の正式結果表示は禁止")
    import pandas as pd
    from IPython.display import display
    print(json.dumps(summary,ensure_ascii=False,indent=2))
    results = pd.read_csv(out/"stage6_validation_results.csv.gz")
    periods = pd.read_csv(out/"stage6_period_results.csv.gz")
    if len(results) != 50 or len(periods) != 150:
        raise RuntimeError("正式結果行数の不一致")
    for status in ("PASS", "FAIL", "INSUFFICIENT_SAMPLE"):
        print(status)
        display(results.loc[results.ValidationStatus == status])
    for period in ("2024", "2025", "Combined"):
        print(period)
        display(periods.loc[periods.Period.astype(str) == period])
    print("2026 Monitor未実行。Portfolio未実行。PASSはlive採用ではありません。")
```

```python
if RUN_STAGE6_FULL and SAVE_OUTPUT_TO_DRIVE:
    import shutil
    from pathlib import Path
    progress = json.loads((Path(OUTPUT_ROOT)/"progress.json").read_text())
    if progress["State"] != "COMPLETE_STAGE6_VALIDATION_ONLY":
        raise RuntimeError("完走後だけ保存できます")
    destination = Path(DRIVE_OUTPUT_ROOT)
    if destination.exists():
        raise FileExistsError("既存保存先は上書きしません。新規directoryを指定してください")
    shutil.copytree(OUTPUT_ROOT,destination)
    print("再開用の出力一式を新規Drive保存先へ保存しました")
```
